```python
# import pandas as pd
# from sklearn.preprocessing import LabelEncoder
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv("train.csv")
```

```python
# mode function is return the which come more time use this value in think the value is tied come the 2 value to gether in the mode 
# Example: if values are ['S','C','S','Q','S'], then df["Embarked"].mode() returns ['S'].
# If two values are tied, you get both (e.g. ['C','S']).
df["Embarked"].fillna(df["Embarked"].mode()[0],inplace=True)
# handle the Embarked column 
```

```python
df
```

```python
df["Embarked"].info()
```

```python
df.loc[61,["Name"]]
df.loc[61,["Embarked"]]
```

```python
df["Cabin"].fillna("Unknown", inplace=True)
# handel the cabin data
```

```python
df
```

```python
# handel the gender male and female 
le = LabelEncoder()
df["Sex"] = le.fit_transform(df["Sex"]) # male=1, female=0
```

```python
df
```

```python
# One-Hot Encoding (for nominal categories, no order)
# Creates separate columns for each category value:
df = pd.get_dummies(df, columns=["Embarked"], drop_first=True)
```

```python
df
```

```python
df["Title"] = df["Name"].str.extract(' ([A-Za-z]+)\.', expand=False)
df["Title"] = df["Title"].replace(['Mlle', 'Ms'], 'Miss')
df["Title"] = df["Title"].replace(['Mme'], 'Mrs')

# Encode Title
df = pd.get_dummies(df, columns=["Title"], drop_first=True)
```

```python
df
```

```python
print(df.head())
```

```python
df.drop(columns=["Name", "Ticket", "Cabin"], inplace=True)
```

```python
df
```

```python
df.info()
```

```python
# handel the missing age 
df["Age"].fillna(df["Age"].median(), inplace=True)
df["Fare"].fillna(df["Fare"].median(), inplace=True)
```

```python
df.info()
```

```python
df.drop(columns=["PassengerId"],inplace=True)
```

```python
df
```

```python
# lets take the tabel the from the df
train =  df.drop("Survived",axis=1) # take the only feature 
label = df["Survived"]
```

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(train, label, test_size=0.2, random_state=42)
```

```python
y_train
```

```python
# lets the create the model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
```

```python
predict = model.predict(X_test)
```

```python
predict
```

```python
y_test
```

```python
# Results
# print("Accuracy:", accuracy_score(y_test, predict))
accuracy_percent = accuracy_score(y_test, predict) * 100
print(f"Accuracy: {accuracy_percent:.2f}%")

print("\nClassification Report:\n", classification_report(y_test, predict))
```
